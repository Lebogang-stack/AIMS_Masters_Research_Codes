```python
import os
import re
from datetime import date, timedelta

# Folder containing your CHIRPS files
chirps_download_folder = r"C:\Users\franc\CHIRPS2014"

# Year you want to check
year = 2014

# Get all files in the folder
files = os.listdir(chirps_download_folder)

# Extract dates from filenames
found_dates = set()

for f in files:
    # Looks for dates like 1997.01.01, 1997-01-01, or 19970101
    match = re.search(rf"{year}[._-]?(\d{{2}})[._-]?(\d{{2}})", f)

    if match:
        month = int(match.group(1))
        day = int(match.group(2))

        try:
            found_dates.add(date(year, month, day))
        except ValueError:
            pass

# Generate all expected dates for the year
start_date = date(year, 1, 1)
end_date = date(year, 12, 31)

expected_dates = set()
current_date = start_date

while current_date <= end_date:
    expected_dates.add(current_date)
    current_date += timedelta(days=1)

# Find missing dates
missing_dates = sorted(expected_dates - found_dates)

# Results
print(f"Year: {year}")
print(f"Expected files: {len(expected_dates)}")
print(f"Files containing valid dates: {len(found_dates)}")
print(f"Missing files/dates: {len(missing_dates)}")

if missing_dates:
    print("\nMissing dates:")
    for d in missing_dates:
        print(d.strftime("%Y-%m-%d"))
else:
    print("\n✅ No dates are missing!")
```

```python
import os
import glob
import gzip
import shutil

# --------------------------------------------------
# CHIRPS 1982 EXTRACTION
# --------------------------------------------------

year = 2014

input_folder = rf"C:\Users\franc\CHIRPS{year}"
output_folder = rf"C:\Users\franc\CHIRPS{year}_extracted"

# Create output folder
os.makedirs(output_folder, exist_ok=True)

# Find all .tif.gz files
gz_files = sorted(
    glob.glob(os.path.join(input_folder, "*.tif.gz"))
)

print("Number of .tif.gz files:", len(gz_files))

# Extract each file
for i, gz_file in enumerate(gz_files, start=1):

    # Remove .gz
    tif_name = os.path.basename(gz_file).replace(".gz", "")
    tif_file = os.path.join(output_folder, tif_name)

    # Skip if already extracted
    if not os.path.exists(tif_file):
        with gzip.open(gz_file, "rb") as f_in:
            with open(tif_file, "wb") as f_out:
                shutil.copyfileobj(f_in, f_out)

    if i % 25 == 0 or i == len(gz_files):
        print(f"Extracted {i}/{len(gz_files)}")

print("\nExtraction complete!")

# Check extracted files
tif_files = sorted(
    glob.glob(os.path.join(output_folder, "*.tif"))
)

print("Number of .tif files:", len(tif_files))

if len(tif_files) > 0:
    print("\nFirst file:")
    print(os.path.basename(tif_files[0]))

    print("\nLast file:")
    print(os.path.basename(tif_files[-1]))
```

```python
# ============================================================
# CHIRPS + GLEAM WATER BALANCE FOR ONE YEAR
# Automatically handles leap years
# ============================================================

import os
import glob
import re
import numpy as np
import xarray as xr
import rasterio


# ============================================================
# 1. YEAR
# ============================================================

year = 2014


# ============================================================
# 2. CHECK LEAP YEAR
# ============================================================

is_leap_year = (
    year % 4 == 0
    and (year % 100 != 0 or year % 400 == 0)
)

expected_days = 366 if is_leap_year else 365

print("=" * 60)
print(f"WATER BALANCE FOR {year}")
print("=" * 60)

print("\nYear information:")
print("Leap year:", is_leap_year)
print("Expected days:", expected_days)


# ============================================================
# 3. FOLDERS
# ============================================================

chirps_folder = rf"C:\Users\franc\CHIRPS{year}_extracted"

gleam_file = rf"C:\Users\franc\{year}\Ep_{year}_GLEAM_v4.3a.nc"

water_balance_folder = rf"C:\Users\franc\WaterBalance_{year}"

os.makedirs(
    water_balance_folder,
    exist_ok=True
)


# ============================================================
# 4. CHECK FILES
# ============================================================

if not os.path.exists(chirps_folder):

    raise FileNotFoundError(
        f"CHIRPS folder not found:\n{chirps_folder}"
    )

if not os.path.exists(gleam_file):

    raise FileNotFoundError(
        f"GLEAM file not found:\n{gleam_file}"
    )


# ============================================================
# 5. FIND CHIRPS FILES
# ============================================================

chirps_files = sorted(
    glob.glob(
        os.path.join(
            chirps_folder,
            "*.tif"
        )
    )
)

print(
    "\nNumber of CHIRPS files:",
    len(chirps_files)
)


if len(chirps_files) != expected_days:

    raise ValueError(
        f"Expected {expected_days} CHIRPS files "
        f"for {year}, but found {len(chirps_files)}."
    )


# ============================================================
# 6. TARGET SOUTH AFRICA GRID
# ============================================================

target_lat = np.arange(
    -20.95,
    -34.95 - 0.001,
    -0.1
)

target_lon = np.arange(
    11.05,
    32.95 + 0.001,
    0.1
)

print("\nTarget grid:")

print(
    "Latitude:",
    len(target_lat)
)

print(
    "Longitude:",
    len(target_lon)
)


# ============================================================
# 7. READ CHIRPS DAILY FILES
# ============================================================

daily_chirps = []

dates = []


for i, chirps_file in enumerate(chirps_files):

    filename = os.path.basename(
        chirps_file
    )


    # --------------------------------------------------------
    # Extract date
    # --------------------------------------------------------

    match = re.search(
        rf"{year}\.(\d{{2}})\.(\d{{2}})",
        filename
    )


    if match is None:

        print(
            "Skipping file with unknown date:",
            filename
        )

        continue


    month = int(
        match.group(1)
    )

    day = int(
        match.group(2)
    )


    date = np.datetime64(
        f"{year}-{month:02d}-{day:02d}"
    )


    # --------------------------------------------------------
    # Read TIFF
    # --------------------------------------------------------

    with rasterio.open(
        chirps_file
    ) as src:

        data = src.read(
            1
        ).astype(
            np.float32
        )

        transform = src.transform

        height, width = data.shape


        # ----------------------------------------------------
        # Longitude
        # ----------------------------------------------------

        lon = (
            transform.c
            +
            (np.arange(width) + 0.5)
            *
            transform.a
        )


        # ----------------------------------------------------
        # Latitude
        # ----------------------------------------------------

        lat = (
            transform.f
            +
            (np.arange(height) + 0.5)
            *
            transform.e
        )


    # --------------------------------------------------------
    # Remove CHIRPS missing values
    # --------------------------------------------------------

    data[data <= -9990] = np.nan


    # --------------------------------------------------------
    # Create DataArray
    # --------------------------------------------------------

    da = xr.DataArray(

        data,

        dims=(
            "lat",
            "lon"
        ),

        coords={
            "lat": lat,
            "lon": lon
        },

        name="precipitation"
    )


    # --------------------------------------------------------
    # Sort coordinates
    # --------------------------------------------------------

    if da.lat[0] > da.lat[-1]:

        da = da.sortby(
            "lat"
        )


    if da.lon[0] > da.lon[-1]:

        da = da.sortby(
            "lon"
        )


    # --------------------------------------------------------
    # Select South Africa
    # --------------------------------------------------------

    da = da.sel(

        lat=slice(
            -34.95,
            -20.95
        ),

        lon=slice(
            11.05,
            32.95
        )
    )


    # --------------------------------------------------------
    # Interpolate onto exact grid
    # --------------------------------------------------------

    da = da.interp(

        lat=target_lat[::-1],

        lon=target_lon,

        method="linear"
    )


    # --------------------------------------------------------
    # Make latitude descending
    # --------------------------------------------------------

    da = da.sortby(
        "lat",
        ascending=False
    )


    daily_chirps.append(
        da
    )

    dates.append(
        date
    )


    if (
        (i + 1) % 25 == 0
        or
        (i + 1) == len(chirps_files)
    ):

        print(
            f"Processed {i + 1}/{len(chirps_files)}"
        )


# ============================================================
# 8. CHECK NUMBER OF READ FILES
# ============================================================

if len(daily_chirps) != expected_days:

    raise ValueError(
        f"Only {len(daily_chirps)} CHIRPS files "
        f"were successfully read. "
        f"Expected {expected_days}."
    )


# ============================================================
# 9. COMBINE DAILY CHIRPS
# ============================================================

chirps = xr.concat(
    daily_chirps,
    dim="time"
)

chirps = chirps.assign_coords(
    time=(
        "time",
        dates
    )
)

chirps = chirps.sortby(
    "time"
)


# ============================================================
# 10. REMOVE DUPLICATE CHIRPS DATES IF NECESSARY
# ============================================================

unique_dates, unique_indices = np.unique(
    chirps.time.values,
    return_index=True
)

if len(unique_dates) != chirps.sizes["time"]:

    print(
        "\nDuplicate CHIRPS dates detected."
    )

    unique_indices = np.sort(
        unique_indices
    )

    chirps = chirps.isel(
        time=unique_indices
    )


# ============================================================
# 11. CHIRPS CHECK
# ============================================================

print("\n" + "=" * 60)
print(f"CHIRPS {year}")
print("=" * 60)

print(
    "Shape:",
    chirps.shape
)

print(
    "First date:",
    chirps.time.values[0]
)

print(
    "Last date:",
    chirps.time.values[-1]
)

print(
    "Number of days:",
    chirps.sizes["time"]
)

print(
    "Latitude:",
    float(chirps.lat.min()),
    "to",
    float(chirps.lat.max())
)

print(
    "Longitude:",
    float(chirps.lon.min()),
    "to",
    float(chirps.lon.max())
)

print(
    "Minimum:",
    float(chirps.min().values)
)

print(
    "Maximum:",
    float(chirps.max().values)
)

print(
    "Mean:",
    float(chirps.mean().values)
)

print(
    "Missing values:",
    int(
        chirps.isnull().sum().values
    )
)


if chirps.sizes["time"] != expected_days:

    raise ValueError(
        f"CHIRPS has {chirps.sizes['time']} days "
        f"instead of expected {expected_days}."
    )


if chirps.sizes["lat"] != 141:

    raise ValueError(
        "CHIRPS does not have 141 latitude points."
    )


if chirps.sizes["lon"] != 220:

    raise ValueError(
        "CHIRPS does not have 220 longitude points."
    )


print(
    "\nCHIRPS checks passed!"
)


# ============================================================
# 12. OPEN GLEAM
# ============================================================

print("\n" + "=" * 60)
print("OPENING GLEAM")
print("=" * 60)


ds_gleam = xr.open_dataset(
    gleam_file
)


if "Ep" not in ds_gleam:

    raise ValueError(
        "Ep variable was not found in GLEAM."
    )


gleam_ep = ds_gleam["Ep"]


print(
    "Original GLEAM shape:",
    gleam_ep.shape
)


# ============================================================
# 13. CHECK GLEAM TIME
# ============================================================

print(
    "\nGLEAM first date:",
    gleam_ep.time.values[0]
)

print(
    "GLEAM last date:",
    gleam_ep.time.values[-1]
)

print(
    "GLEAM days:",
    gleam_ep.sizes["time"]
)


if gleam_ep.sizes["time"] != expected_days:

    raise ValueError(
        f"GLEAM has {gleam_ep.sizes['time']} days "
        f"but {expected_days} were expected for {year}."
    )


# ============================================================
# 14. SELECT SOUTH AFRICA
# ============================================================

gleam_ep = gleam_ep.sel(

    lat=slice(
        -20.95,
        -34.95
    ),

    lon=slice(
        11.05,
        32.95
    )
)


# ============================================================
# 15. SORT GLEAM
# ============================================================

gleam_ep = gleam_ep.sortby(
    "lat",
    ascending=False
)

gleam_ep = gleam_ep.sortby(
    "lon"
)


# ============================================================
# 16. ALIGN GLEAM TO CHIRPS GRID
# ============================================================

gleam_ep = gleam_ep.interp(

    lat=chirps.lat,

    lon=chirps.lon,

    method="linear"
)


print(
    "\nAfter spatial alignment:"
)

print(
    "CHIRPS:",
    chirps.shape
)

print(
    "GLEAM:",
    gleam_ep.shape
)


# ============================================================
# 17. CHECK GLEAM TIME DUPLICATES
# ============================================================

gleam_unique_dates = np.unique(
    gleam_ep.time.values
)


if len(gleam_unique_dates) != gleam_ep.sizes["time"]:

    print(
        "\nDuplicate GLEAM dates detected."
    )

    _, unique_indices = np.unique(
        gleam_ep.time.values,
        return_index=True
    )

    unique_indices = np.sort(
        unique_indices
    )

    gleam_ep = gleam_ep.isel(
        time=unique_indices
    )


# ============================================================
# 18. FINAL TIME VALIDATION
# ============================================================

print(
    "\nFinal time check:"
)

print(
    "CHIRPS:",
    chirps.sizes["time"]
)

print(
    "GLEAM:",
    gleam_ep.sizes["time"]
)


if (
    chirps.sizes["time"]
    !=
    gleam_ep.sizes["time"]
):

    raise ValueError(
        "CHIRPS and GLEAM have different "
        "numbers of time steps."
    )


# ============================================================
# 19. ALIGN TIME USING ACTUAL DATES
# ============================================================

chirps_dates = chirps.time.values

gleam_dates = gleam_ep.time.values


if not np.array_equal(
    chirps_dates,
    gleam_dates
):

    print(
        "\nTime coordinates differ."
    )

    print(
        "Aligning GLEAM to CHIRPS dates..."
    )

    gleam_ep = gleam_ep.assign_coords(
        time=chirps_dates
    )


# ============================================================
# 20. FINAL TIME CHECK
# ============================================================

if not np.array_equal(
    chirps.time.values,
    gleam_ep.time.values
):

    raise ValueError(
        "CHIRPS and GLEAM time coordinates "
        "are still not identical."
    )


print(
    "\nTime coordinates match!"
)


# ============================================================
# 21. CALCULATE WATER BALANCE
# ============================================================

water_balance = (
    chirps
    -
    gleam_ep
)


water_balance.name = (
    "water_balance"
)


water_balance.attrs[
    "units"
] = "mm/day"


water_balance.attrs[
    "long_name"
] = (
    "Daily climatic water balance (P - Ep)"
)


# ============================================================
# 22. WATER BALANCE CHECK
# ============================================================

print("\n" + "=" * 60)
print(f"WATER BALANCE {year}")
print("=" * 60)

print(
    "Shape:",
    water_balance.shape
)

print(
    "First date:",
    water_balance.time.values[0]
)

print(
    "Last date:",
    water_balance.time.values[-1]
)

print(
    "Number of days:",
    water_balance.sizes["time"]
)

print(
    "Minimum:",
    float(
        water_balance.min().values
    )
)

print(
    "Maximum:",
    float(
        water_balance.max().values
    )
)

print(
    "Mean:",
    float(
        water_balance.mean().values
    )
)

print(
    "Missing values:",
    int(
        water_balance.isnull().sum().values
    )
)


# ============================================================
# 23. TEST LOCATION
# ============================================================

lat_test = -27.95
lon_test = 21.05


wb_test = water_balance.sel(

    lat=lat_test,

    lon=lon_test,

    method="nearest"
)


print("\n" + "=" * 60)
print("TEST LOCATION")
print("=" * 60)

print(
    "Latitude:",
    float(wb_test.lat)
)

print(
    "Longitude:",
    float(wb_test.lon)
)

print(
    "\nFirst 10 daily P - Ep values:"
)

print(
    wb_test.values[:10]
)

print(
    "\nMinimum:",
    float(
        wb_test.min().values
    )
)

print(
    "Maximum:",
    float(
        wb_test.max().values
    )
)

print(
    "Mean:",
    float(
        wb_test.mean().values
    )
)

print(
    "Missing:",
    int(
        wb_test.isnull().sum().values
    )
)


# ============================================================
# 24. SAVE WATER BALANCE
# ============================================================

output_file = os.path.join(

    water_balance_folder,

    f"water_balance_{year}.nc"
)


water_balance.to_netcdf(
    output_file
)


print(
    "\nSaved:"
)

print(
    output_file
)


print(
    "\n" + "=" * 60
)

print(
    "DONE!"
)

print(
    "=" * 60
)
```
